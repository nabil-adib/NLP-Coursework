```python
import re
 
with open("Faculty_of_Science_and_Technology___AIUB.html",
          "r",
          encoding="utf-8",
          errors="ignore") as f:
    html = f.read()
 
name_pattern = r'<h2[^>]*x-text="profile\.CvPersonal\.Name"[^>]*>\s*(.*?)\s*</h2>'
names = list(re.finditer(name_pattern, html, re.S))
 
with open("faculty.csv", "w", encoding="utf-8-sig") as f:
 
    f.write("Faculty Name,Email\n")
 
    for i in range(len(names)):
 
        name = names[i].group(1).strip()
 
        if i < len(names) - 1:
            section = html[names[i].end():names[i+1].start()]
        else:
            section = html[names[i].end():]
 
        email_match = re.search(
            r'#([A-Za-z0-9._%+-]+@aiub\.edu)',
            section
        )
 
        if email_match:
            email = email_match.group(1)
        else:
            email = ""
 
        f.write(f'"{name}","{email}"\n')
 
print("Done")
```
