```python
import re
from pypdf import PdfReader

# Read all the text from the PDF
reader = PdfReader("b-pharm.pdf")
text = ""
for page in reader.pages:
    text += page.extract_text()

# Student ID format: 2 digits - 5 digits - 1 digit (e.g., 27-10002-1)
ids = re.findall(r"\b\d{2}-\d{5}-\d\b", text)

# Print the IDs
for student_id in ids:
    print(student_id)

print("Total IDs found:", len(ids))
```
